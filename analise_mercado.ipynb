### Agentes
- Pesquisador de Mecado 
- Analista de Tendências
- Redator do Relatório

```python
from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv())
```

```python
from crewai import Agent, Task, Crew
```

```python
pesquisador = Agent(
    role="Pesquisador de Mercado",
    goal="Coletar e organizar informações relevantes sobre {sector}",
    backstory="""
Você é um pesquisador experiente que analisa tendências de mercado e coleta
dados relevantes sobre {sector}. Seu trabalho é garantir que todas as informações
estejam atualizadas e bem documentadas.
""",
    allow_delegation=False,
    verbose=True
)
```

```python
analista = Agent(
    role="Analista de Tendências",
    goal="Analisar ps dados dp setor {sector} e identificar padrões e oportunidades",
    backstory="""
    Você é um analista de mercado que examina os dados coletados para identificar tendências
    emergentes, oportunidades e ameaças no setor {sector}
    """,
    allow_delegation=False,
    verbose=True
)
```

```python
redator = Agent(
    role="Redator de Relatórios",
    goal="Elaborar um relatório consolidado sonre a análise de mercado do setor {sector}",
    backstory="""
Você é um redator profissional que transforma análises de mercado em um relatório estruturado e compreensível para tomadores de decisão.data=
""",
    allow_delegation=False,
    verbose=True
)
```

```python
coleta_dados=Task(
    description=(
        "1.Pesquisar e coletar informações atualizadas sobre o setor {sector}."
        "2.Identificar os principais players, tendências e estatísticas do setor."
        "3.Organizar os dados de forma clara para análise."
    ),
    expected_output="Um relatório estruturado contendo dados de mercado sobre {sector}.",
    agent=pesquisador
)
```

```python
analise_tendencias = Task(
    description=(
        "1.Examinar os dados coletados pelo Pesquisador de Mercado."
        "2.Identificar padrões, tendências emergentes e oportunidades do setor {sector}."
        "3.Elaborar umaanálise detalhada destacando os principais pontos."
    ),
    expected_output="Um relatório com insights e tendências baseadas nos dados do setor {sector}.",
    agent=analista
)
```

```python
redacao_relatorio = Task(
    description=(
        "1.usar a análise de tendências para criar um relatório detalhado sobre {sector}."
        "2.Garantir que o relatório seja bem estruturado e compreensivel."
        "3.Apresentar um resumo executivo e recomendações finais."
    ),
    expected_output="Um relatório de análise de mercado em formato Markdown, pronto para leitura e apresentação.",
    agent=redator
)
```

```python
crew = Crew(
    agents = [pesquisador, analista, redator],
    tasks = [coleta_dados, analise_tendencias, redacao_relatorio],
    verbose = True
)
```

```python
crew
```

```python
resultado = crew.kickoff(inputs={"sector": "Inteligência artificial"})
```
